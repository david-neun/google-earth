```python
import sys
import os
from scripts.kml_utils import parse_kml_coordinates
import ee
import geemap
import folium
from colorama import Fore, Back, Style

#my helper here
# parse_kml_coordinates already imported above

# The main idea of this notebook is extract the embeddings
# more info about embeddings here:
# https://developers.google.com/machine-learning/crash-course/embeddings/
# https://developers.google.com/machine-learning/crash-course/embeddings/embedding-space
# something defenitivily worthy to study:
# https://deepmind.google.com/science/weatherlab

# Initialize the Google Earth Engine module in your notebook session before running the cells below.
ee.Authenticate()
# This is my project, to replicate this experiment you should go to google earth engine and copy your Project ID in the line below
ee.Initialize(project='eastern-thinker-471320-h4')

# test start here
print(Back.RED + 'This is a test' + Back.RESET)
embeddings = ee.ImageCollection('GOOGLE/SATELLITE_EMBEDDING/V1/ANNUAL')
black_coords = [
    [-104.2, 44.9],
    [-104.2, 43.0],
    [-103.0, 43.0],
    [-103.0, 44.9],
    [-104.2, 44.9],
]


#filter the embeddings for the black hills region to avoid loading the entire collection and run out of memory
# prepare the date range
year = 2024
startDate = ee.Date.fromYMD(year, 1, 1)
endDate = startDate.advance(1, 'year')
# set the filter by date and region: build a proper EE geometry and use filterBounds
black_geom = ee.Geometry.Polygon(black_coords)
filteredEmbeddings = embeddings.filterDate(startDate, endDate).filterBounds(black_geom)
embeddingsImage = filteredEmbeddings.mosaic()
print('Satellite Embedding Image', embeddingsImage)
print(Back.GREEN + 'Embeddings bands' + Back.RESET , embeddingsImage.bandNames().getInfo())
# Use string keys for visualization params and geemap.Map instead of ee.Map
visParams = {'min': -0.3, 'max': 0.3, 'bands': ['A01', 'A16', 'A09']}
Map = geemap.Map()
Map.centerObject(black_geom, 9)
Map.addLayer(embeddingsImage.clip(black_geom), visParams, 'Embeddings Image', opacity=0.8)
display(Map)

# Well, that was fu :) ... now let's try to do some clusters, and later some classification using 
# the embeddings as features for a supervised classification:
# https://developers.google.com/earth-engine/tutorials/community/satellite-embedding-02-unsupervised-classification
```
